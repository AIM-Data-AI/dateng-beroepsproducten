# Voorbeeld: MSSQL uitlezen met pandas

Deze notebook maakt vanaf je host (dus buiten Docker) verbinding met de
`mssql`-container uit dit voorbeeld en laadt de inhoud van de tabel
`dbo.example_numbers` (gevuld door `app/connect_example.py`) in een
pandas-DataFrame.

> **Let op:** gebruik `127.0.0.1` als host, niet `localhost`. Op sommige
> machines resolvet `localhost` naar het IPv6-adres `::1`, maar op dat IP-adres is de MSSQL-server helemaal niet bereikbaar.

Vereisten:

- `docker compose up --build` moet gedraaid hebben (of nog draaien), zodat de
  tabel bestaat en poort 1433 open staat op `localhost`.
- Lokaal geïnstalleerd: `pyodbc`, `pandas` en de Microsoft ODBC-driver
  (`msodbcsql18`). Zie de readme voor installatie-instructies per platform.

```python
import os

import pandas as pd
import pyodbc

HOST = os.environ.get("MSSQL_HOST", "127.0.0.1")
PORT = os.environ.get("MSSQL_PORT", "1433")
USER = os.environ.get("MSSQL_USER", "sa")
PASSWORD = os.environ.get("MSSQL_SA_PASSWORD", "YourStrong!Passw0rd")
DATABASE = os.environ.get("MSSQL_DATABASE", "master")

CONN_STR = (
    "DRIVER={ODBC Driver 18 for SQL Server};"
    f"SERVER={HOST},{PORT};"
    f"DATABASE={DATABASE};"
    f"UID={USER};PWD={PASSWORD};"
    "TrustServerCertificate=yes;"
)

conn = pyodbc.connect(CONN_STR, timeout=10)
```

```python
df = pd.read_sql("SELECT id, value FROM dbo.example_numbers ORDER BY id;", conn)
df
```
